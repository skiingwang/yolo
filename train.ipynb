```python
def train(model, train_loader, epochs=10, lr=0.0001, s=7, b=2, c=10):
    optimizer = paddle.optimizer.AdamW(parameters=model.parameters(), learning_rate=lr)
    loss_fn = YoloLoss()
    losses = []
    val_losses = []
    since = time.time()
    for epoch in range(epochs):
        model = model.to(device).train()
        print(f'Epoch {epoch+1}/{epochs}')
        batch_losses = []
        for batch_idx, batch in enumerate(train_loader):
            img, labels, boxes = batch
            img = img.to(device)
            logits = model(img)  # [batch_size, s, s, b * 5 + c]
            target = preprocessor(labels, boxes, s, b, c, logits)  # boxes: List, logits: Tensor
            loss = loss_fn(logits, target)
            batch_losses.append(loss)
            optimizer.clear_grad()
            loss.backward()
            optimizer.step()
        losses.append(np.mean(batch_losses))
        print(f'Loss: {losses[-1]:.4f}')
        print('-'*32)
    time_elapsed = time.time() - since
    print(f'Training complete in {time_elapsed//60:.0f}m {time_elapsed%60:.0f}s')
    return losses, val_losses
```

```python
train(model, './yolo1.pdparams', train_loader, val_loader)
```

```python
def calculate_iou(box1, boxes):
    """
    box1: [x1, y1, w1, h1] (绝对坐标，中心点格式)
    boxes: [N, 4] (绝对坐标，中心点格式)
    """
    ious = []

    box1_x, box1_y, box1_w, box1_h = box1

    for box in boxes:
        box_x, box_y, box_w, box_h = box
    # 计算交集
        inter_x1 = paddle.max(paddle.to_tensor([box1_x - box1_w / 2, box_x - box_w / 2]))
        inter_y1 = paddle.max(paddle.to_tensor([box1_y - box1_h / 2, box_y - box_h / 2]))
        inter_x2 = paddle.min(paddle.to_tensor([box1_x + box1_w / 2, box_x + box_w / 2]))
        inter_y2 = paddle.min(paddle.to_tensor([box1_y + box1_h / 2, box_y + box_h / 2]))

        inter_w = paddle.max(paddle.to_tensor([0, inter_x2 - inter_x1]))
        inter_h = paddle.max(paddle.to_tensor([0, inter_y2 - inter_y1]))
        intersection = inter_w * inter_h

        # 计算并集
        area1 = box1[2] * box1[3]
        area2 = boxes[:, 2] * boxes[:, 3]
        union = area1 + area2 - intersection

        iou = intersection / (union + 1e-6)
        ious.append(iou)
    return paddle.max(paddle.to_tensor(ious))
```

```python
def nms(predictions, threshold=0.5):
    if len(predictions) == 0:
        return []
    sorted_idx = paddle.argsort(predictions[:, -1], descending=True)
    predictions = predictions[sorted_idx]

    keep = []

    while len(predictions) > 0:
        current = predictions[0]
        keep.append(current)
        if len(predictions) == 1:
            break
        ious = calculate_iou(current[:4], predictions[:, :4])
        predictions = predictions[1:][ious < threshold]

    return paddle.to_tensor(keep)
```
