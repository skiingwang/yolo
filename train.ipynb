```python
import paddle
from paddle.io import DataLoader
from yolov1 import Yolo1
from dataset import YoloDataset, collate_fn, preprocessor, img_scale, Compose
from utils import save_weight, load_weight
from trainer import train
```

```python
model = Yolo1()
```

```python
paddle.summary(model, (1, 3, 448, 448))
```

```python
transformer = Compose([
    img_scale(448, 448),
    preprocessor
])

ds = YoloDataset('./data/Drink_284_Detection', transform=transformer)
train_loader = DataLoader(ds, batch_size=8, shuffle=True, collate_fn=collate_fn)
val_loader = DataLoader(ds, batch_size=8, shuffle=True, collate_fn=collate_fn)
```

```python
imgs, labels = next(iter(train_loader))
```

```python
import matplotlib.pyplot as plt
plt.imshow(imgs[0].numpy().transpose(2, 1, 0))
imgs.shape
```

```python
mask_obj = labels[0][:, :, 0] == 1
row, col = paddle.where(mask_obj)
a = paddle.to_tensor([*zip(row, col)])
a
```

```python
for i in range(7):
    for j in range(7):
        if labels[0, i, j, 0] == 1:
            cls = paddle.where(labels[0, i, j, 5:15] == 1)
            # 4 0.0711 0.3766 0.1422 0.7532
            print(f'Grid ({i}, {j}): Label-{labels[0, i, j, 1:5]} Cls-{cls}')
```

```python
train(model, train_loader, val_loader)
```
